```python
import json
import matplotlib.pyplot as plt
from collections import Counter

with open("data/alfred_dataset.json") as f:
    raw = json.load(f)

all_todos = [todo for ep in raw["train"] for todo in ep]
tiers = [t["tier"] for t in all_todos]
agents = [t.get("expected_agent", "none") for t in all_todos]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

tier_counts = Counter(tiers)
ax1.bar(tier_counts.keys(), tier_counts.values(), color=["red", "blue", "grey"])
ax1.set_title("Distribution of Task Tiers")

agent_counts = Counter(agents)
ax2.bar(agent_counts.keys(), agent_counts.values(), color="purple")
ax2.set_title("Distribution of Specialized Agents")
plt.xticks(rotation=45)
plt.show()
```

```python
from agents.orchestrator import Orchestrator
orch = Orchestrator()

def check_overlap(episode):
    has_t1 = any(t["tier"] == "TIER1_PERSONAL" for t in episode)
    has_t2 = any(t["tier"] == "TIER2_PROFESSIONAL" for t in episode)
    return has_t1 and has_t2

conflict_episodes = [ep for ep in raw["train"] if check_overlap(ep)]
print(f"Episodes with Priority Conflicts: {len(conflict_episodes)}")
print(f"Percentage of Training Data requiring Alfred Protocol: {(len(conflict_episodes)/len(raw['train']))*100:.2f}%")
```
