```python
import mlflow

mlflow.set_experiment("alfred_hyperparameter_optimization")

sweep_params = {
    "learning_rate": [1e-5, 2e-5, 5e-5],
    "batch_size": [4, 8],
    "lora_rank": [8, 16, 32]
}

print("Status: MLflow Experiment Initialized.")
```

```python
from trl import GRPOConfig, GRPOTrainer
from unsloth import FastLanguageModel

def train_with_params(lr, batch_size, r_val):
    with mlflow.start_run(run_name=f"lr_{lr}_bs_{batch_size}_r_{r_val}"):
        mlflow.log_params({
            "learning_rate": lr,
            "batch_size": batch_size,
            "lora_rank": r_val
        })
        
        training_args = GRPOConfig(
            output_dir="./temp-sweep-output",
            num_train_epochs=1,
            per_device_train_batch_size=batch_size,
            learning_rate=lr,
            max_steps=100, 
            report_to="none"
        )
        
     
        
        trainer = GRPOTrainer(
            model=model,
            args=training_args,
            reward_funcs=alfred_reward_fn,
            train_dataset=train_dataset,
            eval_dataset=eval_dataset,
        )
        
        results = trainer.train()
        
        
        final_reward = trainer.state.log_history[-1].get("reward", 0)
        mlflow.log_metric("final_eval_reward", final_reward)
        
        return final_reward

print("Status: Objective Function Defined.")
```

```python
import itertools


keys, values = zip(*sweep_params.items())
combinations = [dict(zip(keys, v)) for v in itertools.product(*values)]

print(f"Status: Starting optimization for {len(combinations)} combinations.")

for config in combinations:
    print(f"Testing: LR={config['learning_rate']}, BS={config['batch_size']}, R={config['lora_rank']}")
    train_with_params(
        config['learning_rate'], 
        config['batch_size'], 
        config['lora_rank']
    )

print("Status: Parameter optimization complete.")
```

```python
import pandas as pd


runs = mlflow.search_runs(experiment_names=["alfred_hyperparameter_optimization"])
sorted_runs = runs.sort_values("metrics.final_eval_reward", ascending=False)

best_run = sorted_runs.iloc[0]
print("Best Parameters Found:")
print(f"  Learning Rate: {best_run['params.learning_rate']}")
print(f"  Batch Size: {best_run['params.batch_size']}")
print(f"  LoRA Rank: {best_run['params.lora_rank']}")
print(f"  Final Reward: {best_run['metrics.final_eval_reward']}")


best_config = {
    "learning_rate": best_run['params.learning_rate'],
    "batch_size": best_run['params.batch_size'],
    "lora_rank": best_run['params.lora_rank']
}
with open("data/best_alfred_config.json", "w") as f:
    json.dump(best_config, f)
```
